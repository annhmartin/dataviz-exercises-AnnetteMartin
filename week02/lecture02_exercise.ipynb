## Task 2 — Bottom vs. Top: A Contrast Story

**What to build:** A bar chart that highlights the **gap between the happiest and least happy countries**, focusing on a specific insight.

**Requirements:**
- Show the **top 8 AND bottom 8 countries** together (16 bars total)
- Use **colour** to distinguish the two groups (not Plotly's default rainbow)
- Add a **visual separator or annotation** that emphasises the gap
- Insight title that tells the story of the gap

**Hint:** Use `pd.concat([df.nlargest(8,'Happiness_Score'), df.nsmallest(8,'Happiness_Score')])` to get both groups.

**Stretch goal:** Add a vertical reference line showing the global average.


```python
# Task 2: Top 8 vs. Bottom 8 contrast
# ------------------------------------

# Step 1: Get top and bottom countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

# Step 2: Build your chart
# YOUR CODE HERE

```

```python
spacer = pd.DataFrame({
    'Country': [''],
    'Happiness_Score': [np.nan],
    'Group': ['gap']
})

bottom8_sorted = bottom8.sort_values('Happiness_Score', ascending=False) 
top8_sorted    = top8.sort_values('Happiness_Score', ascending=False)
combined = pd.concat([top8_sorted, spacer, bottom8_sorted], ignore_index=True)

fig = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    orientation='h',
    color='Group',
    text='Happiness_Score',
    title='<b>The Happiness Gap: World\'s Best vs. Worst</b><br><sup>Top 8 and Bottom 8 countries by Happiness Score</sup>',
    color_discrete_map={
        'Top 8':    '#108783',
        'Bottom 8': '#8C4830',
        'gap':      'rgba(0,0,0,0)',
    },
    category_orders={
        'Country': combined['Country'].tolist(), 
        'Group':   ['Top 8', 'Bottom 8'], 
    },
)

fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='grey',
    line_width=1.5,
    annotation_text=f'Global avg: {global_avg:.2f}',
    annotation_position='bottom right',
    annotation_font_color='grey',
)

fig.update_layout(
    plot_bgcolor='white',
    xaxis=dict(
        title='Happiness Score',
        showgrid=True,
        gridcolor='lightgrey',
        gridwidth=0.5,
        range=[0, combined['Happiness_Score'].max(skipna=True) * 1.15],
    ),
    yaxis_title=None,
    legend=dict(
        title=None,
        x=1,
        y=0,
        xanchor='right',
        yanchor='bottom',
    ),
    margin=dict(l=20, r=80, t=80, b=40),
)

fig.show()
```
