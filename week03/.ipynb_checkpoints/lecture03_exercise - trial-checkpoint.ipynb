```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

---
## Task 1 — Multi-Series Line Chart with Highlight

**What to build:** A line chart showing CO2 emissions over time for **all Asian countries** in the dataset, with one country highlighted.

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#DDDDDD), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

> 💡 `df[df['Region'] == 'Asia']` to filter; use `go.Figure()` with a loop for per-country control


```python
# Task 1 — Multi-series line with highlight
asia = df[df['Region'] == 'Asia'].copy()
highlight = 'China'
highlight_color = '#E63946'
grey = '#DDDDDD'

fig = go.Figure()

for country in asia['Country'].unique():
    d = asia[asia['Country'] == country]
    if country == highlight:
        fig.add_trace(go.Scatter(
            x=d['Year'], y=d['CO2_Mt'],
            mode='lines',
            line=dict(color=highlight_color, width=3),
            name=country,
            showlegend=False
        ))
        # Direct label at the end of the line
        last = d[d['Year'] == d['Year'].max()].iloc[0]
        fig.add_annotation(
            x=last['Year'], y=last['CO2_Mt'],
            text=f"<b>{country}</b>",
            xanchor='left', showarrow=False,
            xshift=8, font=dict(color=highlight_color, size=12, family='Arial')
        )
    else:
        fig.add_trace(go.Scatter(
            x=d['Year'], y=d['CO2_Mt'],
            mode='lines',
            line=dict(color=grey, width=1.2),
            name=country,
            showlegend=False,
            hoverinfo='skip'
        ))

fig.update_layout(
    title=dict(
        text="China's CO₂ emissions tripled while other Asian nations grew modestly",
        font=dict(family='Arial', size=16, color='#222222'),
        x=0.0, xanchor='left'
    ),
    xaxis=dict(
        title='', showgrid=False, zeroline=False,
        tickfont=dict(family='Arial', size=11, color='#555555')
    ),
    yaxis=dict(
        title='CO₂ Emissions (Mt)',
        showgrid=True, gridcolor='#F0F0F0', zeroline=False,
        tickfont=dict(family='Arial', size=11, color='#555555'),
        title_font=dict(family='Arial', size=12, color='#555555')
    ),
    plot_bgcolor='white',
    paper_bgcolor='white',
    margin=dict(l=60, r=120, t=60, b=50),
    font=dict(family='Arial'),
    width=820, height=460
)

fig.show()
```

## Task 2 — Slopegraph: Regional Change 2000 vs 2022

**What to build:** A slopegraph comparing **average regional CO2 emissions** between 2000 and 2022.

**Requirements:**
- One line per region (not per country — aggregate first)
- Colour: regions that increased = one colour; decreased = another
- Values labelled at both ends of each line
- No y-axis tick labels (the endpoint labels make them redundant)
- Insight title stating which regions moved most

> 💡 `df.groupby(['Region','Year'])['CO2_Mt'].mean().reset_index()` then filter to 2000 and 2022


```python
# Task 2 — Slopegraph: regional averages
regional = df.groupby(['Region', 'Year'])['CO2_Mt'].mean().reset_index()
r2000 = regional[regional['Year'] == 2000].set_index('Region')['CO2_Mt']
r2022 = regional[regional['Year'] == 2022].set_index('Region')['CO2_Mt']

regions = r2000.index.tolist()
increased_color = '#E63946'   # red — went up
decreased_color = '#457B9D'   # blue — went down

fig = go.Figure()

for region in regions:
    v0 = r2000[region]
    v1 = r2022[region]
    color = increased_color if v1 > v0 else decreased_color

    fig.add_trace(go.Scatter(
        x=[2000, 2022], y=[v0, v1],
        mode='lines+markers',
        line=dict(color=color, width=2),
        marker=dict(size=7, color=color),
        name=region,
        showlegend=False
    ))

    # Left label
    fig.add_annotation(
        x=2000, y=v0,
        text=f"{region}  {v0:,.0f}",
        xanchor='right', showarrow=False,
        xshift=-8, font=dict(family='Arial', size=11, color=color)
    )
    # Right label
    fig.add_annotation(
        x=2022, y=v1,
        text=f"{v1:,.0f}  {region}",
        xanchor='left', showarrow=False,
        xshift=8, font=dict(family='Arial', size=11, color=color)
    )

fig.update_layout(
    title=dict(
        text="Asia's average CO₂ soared; Europe and Oceania quietly declined 2000–2022",
        font=dict(family='Arial', size=16, color='#222222'),
        x=0.0, xanchor='left'
    ),
    xaxis=dict(
        tickvals=[2000, 2022],
        ticktext=['2000', '2022'],
        showgrid=False, zeroline=False,
        tickfont=dict(family='Arial', size=13, color='#333333')
    ),
    yaxis=dict(
        showticklabels=False,
        showgrid=False, zeroline=False
    ),
    plot_bgcolor='white',
    paper_bgcolor='white',
    margin=dict(l=180, r=180, t=70, b=50),
    font=dict(family='Arial'),
    width=820, height=480
)

fig.show()
```
